## 1. Импорт библиотек и воспроизводимость

pandas используется для работы с CSV-таблицами, NumPy — для числовых проверок, а scikit-learn предоставляет модель и метрику ROC-AUC. PyTorch не используется, поскольку предоставленные файлы уже содержат необходимые 512 числовых признаков.

```python
# DATA_DIR — каталог, содержащий четыре предоставленных CSV-файла.
DATA_DIR = Path()

train = pd.read_csv(DATA_DIR / "train.csv")
validation = pd.read_csv(DATA_DIR / "validation.csv")
test = pd.read_csv(DATA_DIR / "test.csv")
sample_submission = pd.read_csv(DATA_DIR / "sample_submission.csv")

print("Обучающая выборка:", train.shape)
print("Валидационная выборка:", validation.shape)
print("Тестовая выборка:", test.shape)
print("Образец отправки:", sample_submission.shape)
```

## На соревнование:

1) в риадми говорят что это визуальные данные -> это видимо вектора какой-то модели которой скормили картинки и наша задача по этим векторам понять 2 картинки одинаовые или нит

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Разделяем исходные 512 колонок на сущность u и сущность v
feature_cols = [c for c in train.columns if c not in {"row_id", "target"}]
u = train[feature_cols[:256]].values
v = train[feature_cols[256:]].values
y = train["target"].values

# 1. Моменты по признакам (среднее и std для каждой из 256 координат)
mean_u, mean_v = u.mean(axis=0), v.mean(axis=0)
std_u, std_v = u.std(axis=0), v.std(axis=0)

# 2. Евклидовы нормы (длины векторов объектов)
norm_u = np.linalg.norm(u, axis=1)
norm_v = np.linalg.norm(v, axis=1)

# 3. Евклидово расстояние между парами в строке
diff_norm = np.linalg.norm(u - v, axis=1)

# Отрисовка графиков
plt.figure(figsize=(16, 10))

# График 1: Средние значения признаков
plt.subplot(2, 2, 1)
sns.kdeplot(
    mean_u,
    label="Сущность u (0..255)",
    fill=True,
    color="#1f77b4",
    alpha=0.4,
)
sns.kdeplot(
    mean_v,
    label="Сущность v (256..511)",
    fill=True,
    color="#ff7f0e",
    alpha=0.4,
)
plt.title("1. Распределение средних значений признаков", fontsize=12, fontweight="bold")
plt.xlabel("Среднее значение признака")
plt.ylabel("Плотность")
plt.legend()
plt.grid(True, alpha=0.3)

# График 2: Разброс признаков
plt.subplot(2, 2, 2)
sns.kdeplot(
    std_u,
    label="Сущность u (std)",
    fill=True,
    color="#1f77b4",
    alpha=0.4,
)
sns.kdeplot(
    std_v,
    label="Сущность v (std)",
    fill=True,
    color="#ff7f0e",
    alpha=0.4,
)
plt.title(
    "2. Распределение разброса признаков (Std)",
    fontsize=12,
    fontweight="bold",
)
plt.xlabel("Стандартное отклонение признака")
plt.ylabel("Плотность")
plt.legend()
plt.grid(True, alpha=0.3)

# График 3: L2-нормы векторов объектов
plt.subplot(2, 2, 3)
sns.kdeplot(
    norm_u,
    label="Норма ||u|| (Сущность 1)",
    fill=True,
    color="#2ca02c",
    alpha=0.4,
)
sns.kdeplot(
    norm_v,
    label="Норма ||v|| (Сущность 2)",
    fill=True,
    color="#9467bd",
    alpha=0.4,
)
plt.title(
    "3. Распределение длины векторов объектов (L2-норма)",
    fontsize=12,
    fontweight="bold",
)
plt.xlabel("L2-норма вектора ||x||_2")
plt.ylabel("Плотность")
plt.legend()
plt.grid(True, alpha=0.3)

# График 4: Разделение классов по расстоянию ||u - v||
plt.subplot(2, 2, 4)
sns.kdeplot(
    diff_norm[y == 1],
    label="Совпадают (Target = 1)",
    fill=True,
    color="#2ca02c",
    alpha=0.4,
)
sns.kdeplot(
    diff_norm[y == 0],
    label="Не совпадают (Target = 0)",
    fill=True,
    color="#d62728",
    alpha=0.4,
)
plt.title(
    "4. Евклидово расстояние ||u - v|| по целевым классам",
    fontsize=12,
    fontweight="bold",
)
plt.xlabel("Расстояние ||u - v||_2")
plt.ylabel("Плотность")
plt.legend()
plt.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

# Текстовые статистики
print(f"Средняя норма u: {norm_u.mean():.2f} ± {norm_u.std():.2f}")
print(f"Средняя норма v: {norm_v.mean():.2f} ± {norm_v.std():.2f}")
print(f"Среднее расстояние при Target=1: {diff_norm[y == 1].mean():.2f}")
print(f"Среднее расстояние при Target=0: {diff_norm[y == 0].mean():.2f}")
```
